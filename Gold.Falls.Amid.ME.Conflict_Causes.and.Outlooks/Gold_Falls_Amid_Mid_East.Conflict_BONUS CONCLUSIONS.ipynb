# **Gold Price Forecast 2026: What the Experts Are Saying**
### The market is rocking, but interest in the precious metal remains high

At the end of January 2026, gold prices reached an all-time high and then sharply declined.

Back in the fall of 2025, we asked investment analysts whether private investors should add precious metals to their portfolios. In February 2026, we asked experts to update their forecasts and assess gold's prospects for the remainder of the year. We also inquired about the factors that will influence the price of the metal, which **Bloomberg** recently described as a more volatile asset than Bitcoin.

> ### **Growth will likely resume after a healthy correction**
>
> After hitting historic highs of $5,600 per ounce, exchange prices for gold collapsed by more than 20% from their peak values—dropping to $4,400 per ounce.
>
> **Why volatility emerged in the gold market:** Corrective movements in commodities markets were long overdue. Gold accelerated significantly, following a parabolic upward trajectory with almost no breaks or substantial pullbacks, gaining nearly 30% at its peak since the start of 2026. Such a surge was purely speculative, and the overheating of the gold market reached extreme levels.
>
> The trigger for the correction was the news that Trump nominated **Kevin Warsh**, a former member of the Fed Board of Governors, as the new Chairman of the US Federal Reserve. He is seen as the most conservative and "hawkish" candidate among those considered by Trump, which eased investor fears regarding the Fed's independence, the risks of rapid rate cuts, and a new wave of inflation in the US.
>
> Against this backdrop, the dollar began a slight recovery after a prolonged decline, further supported by the Fed's decision to maintain interest rates at the January 28 meeting. Investors moved to take profits in gold, causing a rather sharp correction.
>
> Volatility in precious metals, fueled by investment and speculative demand, is becoming extreme. In the short term, we do not rule out that corrective movements may continue. The next support zone from current levels will be the **$4,400–$4,600** range. However, the fundamental factors remain:
> * Geopolitical and economic tensions;
> * Fed monetary policy easing;
> * Instability and uncertainty in US Treasuries (UST);
> * A crisis of confidence in the US administration and American assets.
>
> **What the gold price will be and the factors influencing it:** The **World Gold Council (WGC)** expects 2026 to be another year of active gold purchases by central banks. However, a downward trend in volume is possible: WGC estimates that volumes could be **200–250 tons less** than in 2025.
>
> Amid high prices and the rising value of existing reserves last year, central banks already reduced gold purchases by 20% to 863.3 tons. However, investment demand for gold in 2025 surged by 84%, reaching 2,175 tons, completely offsetting the decline in demand from regulators.
>
> In 2026, steady capital inflows into gold ETFs and robust demand for bars and coins—driven by persistent economic and geopolitical uncertainty—are expected to sustain general investor interest.
>
> Therefore, after a healthy correction, gold price growth will likely resume. Local declines should be used to cautiously build new positions. The ultra-fast rally at the end of 2025 and beginning of 2026 forced investment firms to revise their forecasts upward—averaging **$6,000–$6,200** per ounce by the end of the year.
>
> We have also raised our 2026 expectations to **$5,500–$5,700** per ounce. Volatility will intensify due to high speculative demand, and profit-taking corrections are inevitable.
>
> On a **2-3 year horizon**, a decrease in geopolitical tension, a stronger dollar, a softening of Trump’s tariffs, or a peace agreement between Russia and Ukraine could lead to a significant drop in gold prices—down to **$3,500–$4,000** per ounce. This would return the gold market closer to fundamental supply and demand valuations.
>
> **Regarding position sizing:** If an investor has no gold positions at all, they can form a long-term position (5+ years) by "laddering" into entries during pullbacks in the **$4,000–$4,500** zone, aiming for **10–15%** of the portfolio. This helps balance risks. If a gold portfolio is already formed, profit-taking should begin at **$5,000–$5,500**, maintaining a volume of no more than **10–15%** of the total portfolio.

> ### **The speculative segment introduces uncertainty into price dynamics**
>
> The key drivers of gold growth are the significantly increased demand for hedging (including against currency and dollar risks) and high speculative interest fueled by expectations of a continued uptrend.
>
> Our base case for 2026 is the persistence of high uncertainty in geopolitics and the global economy. In such an environment, central banks will likely continue to increase gold purchases, and investors will gradually increase its share in long-term portfolios. Therefore, we expect growth to continue after the correction in the medium term (6 months). Although the speculative segment—which has expanded over the past year—introduces significant uncertainty in short-term price dynamics, long-term drivers remain strong.
>
> The factor of declining interest rates remains important but is gradually giving way to a more fundamental theme: the risks of declining trust in the dollar and monetary policy institutions. This could become a more powerful driver for gold than the rate level itself.
>
> Overall, in 2026, demand for gold as a safe-haven asset is likely to dominate. This will compensate for weak industrial demand and smooth out the instability of the speculative component. Consequently, we expect high global demand to persist and see continued price growth throughout the year.
>
> However, it is important to note: the current demand structure and market dynamics suggest increased volatility. Therefore, even with a positive outlook on gold for the year, we do not recommend increasing its share in the portfolio above **10%**.

> ### **Gold's growth factors remain unchanged**
>
> We believe that in the coming months, gold is likely to consolidate within a wide range of **$4,000–$5,000** per ounce. By the end of the year, we expect gold to return to the **$5,500–$6,000** range. In the medium term, we anticipate further price growth based on the same support factors as before.
>
> **Main growth factors for precious metals in recent quarters:**
> * Excess global liquidity;
> * Deglobalization and trade wars;
> * Expectations of further Fed rate cuts;
> * Risks of the Fed losing independence due to pressure from the Trump administration;
> * Rapid growth of US national debt;
> * Political uncertainty in the US.
>
> The price of gold rose rapidly, and a significant part of the final surge was speculative. Therefore, the current cooling is a natural result of overheating. The nomination of Kevin Warsh as the new Fed Chairman candidate served as the catalyst for the correction.
>
> Investor fears of the Fed's independence being undermined, rapid rate cuts, and a new inflationary spiral in the US have slightly decreased. As a result, the global dollar stopped its decline: the **DXY Index** rose from 95.7 to 97 points.
>
> **What should a private investor do?** Increase the gold share or divest? In global practice, an optimal gold allocation is considered to be **5%**. We find this to be reasonable.
